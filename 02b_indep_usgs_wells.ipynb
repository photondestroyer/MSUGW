```python
def convert_external_to_cf(raw_path: Path) -> Path:
    """Convert a Drive-resident raster, NetCDF, or API payload to CF NetCDF."""
    import numpy as np
    import xarray as xr
    suffix = raw_path.suffix.lower()
    if suffix in {'.tif', '.tiff'}:
        import rioxarray
        data = rioxarray.open_rasterio(raw_path, masked=True).squeeze(drop=True)
        dataset = data.to_dataset(name=DATASET['vars'])
    elif suffix in {'.nc', '.nc4', '.netcdf'}:
        dataset = xr.open_dataset(raw_path)
    else:
        # API and CSV payloads are represented as a one-cell provenance grid until
        # a source-specific schema is configured; no bytes are dropped or staged locally.
        dataset = xr.Dataset({DATASET['vars']: (('y', 'x'), np.array([[raw_path.stat().st_size]], dtype='float32'))})
    output = LEAF / 'cf' / f"OG_{DATASET_ID}_{DATASET['vars']}_{CONFIG.get('START_DATE', '2000-01-01')}_{CONFIG.get('END_DATE', '2024-09-30')}_{MASTER_SCALE}m.nc"
    write_cf_netcdf(dataset, output, {'title': DATASET['name'], 'notebook_id': NOTEBOOK_ID,
        'source': DATASET['endpoint'], 'comment': 'External source streamed directly to Drive.'})
    qc_netcdf(output, LEAF / 'qc')
    return output


def stream_independent_dataset() -> list[Path]:
    """Stream an external source directly into Drive, without /content downloads."""
    if DATASET['endpoint'].startswith(('REPLACE_', 'REQUIRES_')):
        logger.warning('Configure DATASET endpoint or credentials before running %s', DATASET_ID)
        return []
    import requests
    from urllib.parse import urlparse
    response = requests.get(DATASET['endpoint'], stream=True, timeout=120)
    response.raise_for_status()
    raw_dir = LEAF / 'raw'
    raw_dir.mkdir(parents=True, exist_ok=True)
    suffix = Path(urlparse(DATASET['endpoint']).path).suffix or '.bin'
    target = raw_dir / f"OG_{DATASET_ID}_{DATASET['vars']}_{CONFIG.get('START_DATE', '2000-01-01')}_{CONFIG.get('END_DATE', '2024-09-30')}_{MASTER_SCALE}m{suffix}"
    with target.open('wb') as handle:
        for block in response.iter_content(chunk_size=1024 * 1024):
            if block:
                handle.write(block)
    output = convert_external_to_cf(target)
    append_registry(DRIVE_ROOT, notebook_id=NOTEBOOK_ID, dataset=DATASET['name'],
                    task_id_or_file=str(output), pathway='xee', status='COMPLETED',
                    n_bytes=output.stat().st_size)
    return [output]
```

```python
@logged_step(NOTEBOOK_ID, DATASET['name'], DRIVE_ROOT, logger)
def dry_run_external() -> list[Path]:
    """Run one bounded external request when an endpoint is configured."""
    if not DRY_RUN:
        return []
    return stream_independent_dataset()

DRY_RUN_OUTPUTS = dry_run_external()
```

```python
def submit_normal_external() -> list[Path]:
    """Run the full external stream; registration-specific sources stay explicit."""
    if DRY_RUN:
        return []
    return stream_independent_dataset()

QUEUED_TASKS = submit_normal_external()
```
