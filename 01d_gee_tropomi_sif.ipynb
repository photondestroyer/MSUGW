```python
DATASET_ID = '01d'
DATASET = {
  "name": "TROPOMI SIF",
  "asset_id": "projects/sat-io/open-datasets/TROPOSIF",
  "vars": "SIF",
  "bands": [
    "SIF"
  ],
  "scale": 1000,
  "kind": "collection",
  "start": "2018-05-01",
  "end": "2024-09-30",
  "folder": "01d_tropomi_sif",
  "usage": "Independent carbon signal for recent-period decoupling analysis."
}
DATASET['drive_subfolder'] = '01_gee_rasters/01d_tropomi_sif'
DATASET['crs'] = 'EPSG:5070'
DATASET['band_metadata'] = {band: {'long_name': band, 'units': '1', 'standard_name': 'unknown'} for band in DATASET['bands']}
```

```python
def build_gee_source() -> tuple[object, bool]:
    """Build clipped, reprojected sources while retaining every native band."""
    asset_ids = DATASET['asset_id'] if isinstance(DATASET['asset_id'], list) else [DATASET['asset_id']]
    if DATASET['kind'] == 'image':
        return ee.Image(asset_ids[0]).clip(AOI), False
    sources = []
    for asset_id in asset_ids:
        collection = ee.ImageCollection(asset_id).filterBounds(AOI).filterDate(DATASET['start'], DATASET['end'])
        # Do not call select(): QA, uncertainty, and native metadata must survive.
        sources.append(collection.map(lambda image: image.clip(AOI).reproject(crs=MASTER_CRS, scale=MASTER_SCALE)))
    return (sources if DATASET['kind'] == 'collection_group' else sources[0]), True

SOURCE, IS_COLLECTION = build_gee_source()
```

```python
# Client-side XEE pulls are disabled in local batch mode.
# Set DRY_RUN only to inspect configuration; it will not create a local NetCDF.
DRY_RUN_OUTPUTS: list[Path] = []
if DRY_RUN:
    logger.info('DRY_RUN is enabled; no Earth Engine export was submitted.')
```

```python
def submit_normal_exports() -> list[str]:
    """Queue temporal or static exports without waiting for server completion."""
    if DRY_RUN:
        return []
    manager = BatchManager(DRIVE_ROOT, NOTEBOOK_ID, DATASET['name'], logger, max_concurrent=20)
    task_ids = []
    folder = f"Ogallala_Phase0/{DATASET['drive_subfolder']}/raw"
    sources = SOURCE if isinstance(SOURCE, list) else [SOURCE]
    for source_index, source in enumerate(sources):
        suffix = f'_part{source_index + 1}' if len(sources) > 1 else ''
        if IS_COLLECTION:
            start_year = int(DATASET['start'][:4])
            end_year = int(DATASET['end'][:4])
            for year in range(start_year, end_year + 1):
                start = f'{year}-01-01'
                end = f'{year + 1}-01-01'
                chunk = source.filterDate(start, end).median().clip(AOI)
                prefix = f"OG_{DATASET_ID}_{DATASET['vars']}{suffix}_{start}_{year}-12-31_{MASTER_SCALE}m"
                task_ids.append(manager.submit_image(chunk, f'{DATASET_ID}_{source_index}_{year}', folder, prefix, AOI, MASTER_SCALE, MASTER_CRS))
        else:
            start, end = DATASET['start'], DATASET['end']
            prefix = f"OG_{DATASET_ID}_{DATASET['vars']}{suffix}_{start}_{end}_{MASTER_SCALE}m"
            task_ids.append(manager.submit_image(source, f'{DATASET_ID}_{source_index}_static', folder, prefix, AOI, MASTER_SCALE, MASTER_CRS))
    return task_ids

QUEUED_TASKS = submit_normal_exports()
```
